```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)  # для воспроизводимости

# Создаем 500 заказов
orders_data = []
for i in range(500):
    orders_data.append({
        'order_id': i + 1,
        'customer_id': np.random.randint(1, 201),
        'product_category': np.random.choice(['Электроника', 'Одежда', 'Книги', 'Красота']),
        'revenue': round(np.random.uniform(20, 300), 2),
        'order_date': pd.date_range('2023-01-01', '2023-12-31', periods=500)[i],
        'country': np.random.choice(['Россия', 'США', 'Германия'])
    })

# Создаем DataFrame
df = pd.DataFrame(orders_data)

# Сохраняем в CSV
df.to_csv('C:/Users/majkl/WorkSpace/Online_store_sales_Analysis/sales_data.csv', index=False, encoding='utf-8')

print(f"Создан файл sales_data.csv с {len(df)} записями")
```
