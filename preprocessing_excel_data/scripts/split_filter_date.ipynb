```python
import pandas as pd
from datetime import datetime

# Укажите путь к вашему файлу Excel
input_file = 'C:/Users/majkl/github_repo/WorkSpace/preprocessing_excel_data/data/Выгурзка_октябрь.xlsx'  # Замените на имя вашего файла
output_file = 'Выгурзка_25.09.25_без_дубликатов.xlsx'  # Имя файла для сохранения результата

# Загрузка данных из Excel
try:
    # Чтение файла (поддержка .xls и .xlsx)
    df = pd.read_excel(input_file)
    
    # Преобразование столбца с датами (если он не автоматически распознан как datetime)
    # Предполагаем, что столбец с датами называется 'Date' - замените на имя вашего столбца
    date_column = 'Дата создания обращения'  # Измените на имя вашего столбца с датами
    df[date_column] = pd.to_datetime(df[date_column], errors='coerce')
    
    # Фильтрация данных: дата >= 18 сентября текущего года
    target_date = datetime(datetime.now().year, 9, 18)  # 18 сентября текущего года
    filtered_df = df[df[date_column] >= target_date]
    
    # Сохранение в новый файл
    filtered_df.to_excel(output_file, index=False)
    print(f"Данные успешно отфильтрованы и сохранены в файл: {output_file}")
    
except Exception as e:
    print(f"Произошла ошибка: {e}")
```

```python
import pandas as pd
import os
from pathlib import Path

def split_by_country(input_file, output_folder):
    """Разделяет Excel-файл по столбцу 'Тип страны' (без chunk-обработки)"""
    try:
        # Читаем весь файл сразу
        df = pd.read_excel(input_file, engine='openpyxl')
        
        # Проверяем наличие столбца "Тип страны"
        if "Тип страны" not in df.columns:
            raise ValueError("❌ Ошибка: Столбец 'Тип страны' не найден!")
        
        # Фильтруем данные (РФ vs другие страны)
        mask_rf = df["Тип страны"].str.contains("РФ|Россия|Russia", case=False, na=False)
        df_rf = df[mask_rf]
        df_other = df[~mask_rf]
        
        # Создаем папку для результатов, если её нет
        os.makedirs(output_folder, exist_ok=True)
        
        # Сохраняем результаты
        base_name = Path(input_file).stem
        df_rf.to_excel(f"{output_folder}/{base_name}_РФ.xlsx", index=False, engine='openpyxl')
        df_other.to_excel(f"{output_folder}/{base_name}_Другие_страны.xlsx", index=False, engine='openpyxl')
        
        print(f"✅ Разделение завершено! Результаты сохранены в: {output_folder}")

    except Exception as e:
        print(f"❌ Ошибка: {e}")

# Укажи путь к файлу и папку для результатов
input_file = "C:/Users/majkl/github_repo/WorkSpace/preprocessing_excel_data/data/Выгурзка_октябрь.xlsx"
output_folder = "split_by_country"

split_by_country(input_file, output_folder)
```
