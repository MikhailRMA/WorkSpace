```python
import pandas as pd
import random

# Загружаем данные
df = pd.read_csv('C:/Users/majkl/WorkSpace/Analisys_seller_mp_oz/anonymize_data/seller_product.csv', sep=';', encoding='utf-8-sig')

# Функции для генерации случайных идентификаторов
def generate_article():
    prefixes = ['AAA', 'BBB', 'CCC', 'DDD', 'EEE', 'FFF', 'GGG', 'HHH']
    materials = ['Black', 'Blue', 'Red', 'Clear', 'Purple', 'Green', 'White', 'Pink', 'Titanium']
    return f"{random.choice(prefixes)}_{random.choice(materials)}_{random.randint(1000, 9999)}"

def generate_product_id():
    return random.randint(10000000, 99999999)

def generate_sku():
    return random.randint(100000000, 999999999)

def generate_barcode():
    # Генерируем 13-значный штрих-код или RND формат
    if random.random() > 0.5:
        return str(random.randint(1000000000000, 9999999999999))
    else:
        return f"RND{random.randint(100000000, 999999999)}"

# Анонимизируем идентификаторы
df['Артикул'] = [generate_article() for _ in range(len(df))]
df['Ozon Product ID'] = [generate_product_id() for _ in range(len(df))]
df['SKU'] = [generate_sku() for _ in range(len(df))]
df['Barcode'] = [generate_barcode() for _ in range(len(df))]

# Сохраняем результат
df.to_csv('anonymous_seller_product.csv', sep=';', index=False, encoding='utf-8-sig')

print("Данные успешно анонимизированы и сохранены в 'anonymous_seller_product.csv'")
print(f"Обработано {len(df)} товаров")
```
