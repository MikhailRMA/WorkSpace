```python
import pandas as pd
import numpy as np
```

```python
#  ГЕНЕРАЦИЯ ДАННЫХ A/B ТЕСТА
np.random.seed(42) # для воспроизводимости результатов

# Параметры генерации
sample_size = 1000 # количество пользователей в каждой группе

# Группа A (старая система рекомендаций)
# Предположим, средний чек = 2500 руб, стандартное отклонение = 800 руб
group_a_spends = np.random.normal(loc=2500, scale=800, size=sample_size)
# Округляем до 2 знаков и убираем возможные отрицательные значения (это же траты)
group_a_spends = np.abs(np.round(group_a_spends, 2))

# Группа B (новая система рекомендаций)
# Предположим, новая система действительно работает и увеличивает средний чек на 150 руб
group_b_spends = np.random.normal(loc=2650, scale=800, size=sample_size)
group_b_spends = np.abs(np.round(group_b_spends, 2))

# Создаем DataFrame
df = pd.DataFrame({
    'user_id': range(1, 2*sample_size + 1),
    'group': ['A'] * sample_size + ['B'] * sample_size,
    'spend': np.concatenate([group_a_spends, group_b_spends])
})
df.to_csv('C:/Users/majkl/WorkSpace/Analysis_AB_test_new_system_on_sales/ab_test_data.csv', index=False, encoding='utf-8')

print(f"Создан файл ab_test_data.csv с {len(df)} записями")
```
