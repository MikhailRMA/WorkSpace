```python
import pandas as pd
import numpy as np
```

```python
#  ГЕНЕРАЦИЯ "ГРЯЗНЫХ" ДАННЫХ
np.random.seed(42) # для воспроизводимости результатов

data = []
for i in range(1000):
    # Создаем реалистичные "грязные" данные
    order_id = i + 1
    
    # Пропуски в customer_id (5% случаев)
    if np.random.random() < 0.05:
        customer_id = None
    else:
        customer_id = np.random.randint(1, 301)
    
    # Разные форматы дат и пропуски
    date_formats = [
        lambda: f"2023-{np.random.randint(1,13):02d}-{np.random.randint(1,29):02d}",  # нормальный
        lambda: f"{np.random.randint(1,13)}/{np.random.randint(1,29)}/2023",  # американский
        lambda: f"{np.random.randint(1,29)}.{np.random.randint(1,13)}.2023",  # европейский
        lambda: "неизвестно",  # текст вместо даты
        lambda: None  # пропуск
    ]
    order_date = np.random.choice(date_formats)()
    
    # Выручка с выбросами и отрицательными значениями
    if np.random.random() < 0.02:  # 2% выбросов
        revenue = np.random.uniform(1000, 5000)  # очень большие заказы
    elif np.random.random() < 0.01:  # 1% отрицательных
        revenue = -np.random.uniform(10, 100)  # возвраты
    elif np.random.random() < 0.03:  # 3% пропусков
        revenue = None
    else:
        revenue = round(np.random.uniform(20, 300), 2)
    
    # Категории с опечатками и разными регистрами
    categories = ['ELECTRONICS', 'electronics', 'Электроника', 'elecronics', 'Eletronics', 
                  'CLOTHING', 'clothing', 'Одежда', 'cloting', None,
                  'BOOKS', 'books', 'Книги', 'bokks', 'BOOK',
                  'BEAUTY', 'beauty', 'Красота', 'beuty', 'BEAUTY ']
    product_category = np.random.choice(categories)
    
    # Страны с разными написаниями
    countries = ['Russia', 'RUSSIA', 'russia', 'Россия', 'RU', 'Russian Federation',
                 'USA', 'United States', 'US', 'America', 'США',
                 'Germany', 'GERMANY', 'germany', 'Германия', 'DE', None]
    country = np.random.choice(countries)
    
    # Email с ошибками
    emails = [f"user{np.random.randint(1,1000)}@gmail.com", 
              f"user{np.random.randint(1,1000)}gmail.com",  # нет @
              f"user{np.random.randint(1,1000)}@",  # нет домена
              "invalid_email",
              None]
    email = np.random.choice(emails)
    
    # Возраст с нереалистичными значениями
    if np.random.random() < 0.01:  # 1% нереалистичных возрастов
        age = np.random.choice([-5, 0, 150, 200])
    elif np.random.random() < 0.04:  # 4% пропусков
        age = None
    else:
        age = np.random.randint(18, 70)
    
    data.append({
        'order_id': order_id,
        'customer_id': customer_id,
        'order_date': order_date,
        'revenue': revenue,
        'product_category': product_category,
        'country': country,
        'email': email,
        'age': age
    })

df = pd.DataFrame(data)

df.to_csv('C:/Users/majkl/WorkSpace/dirty_data_analysis/data.csv', index=False, encoding='utf-8')

print(f" Создано {len(df)} строк с 'грязными' данными")
```
