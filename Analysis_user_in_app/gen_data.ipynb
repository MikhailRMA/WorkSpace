```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta 
```

```python
# Устанавливаем seed для воспроизводимости
np.random.seed(42)

# Создаем список из 1000 условных пользователей
user_ids = [f'user_{i}' for i in range(1000)]

# Генерируем дату регистрации для каждого пользователя за последние 3 месяца
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 4, 1)
registration_dates = []
for i in range(1000):
    random_date = start_date + timedelta(days=np.random.randint(0, (end_date - start_date).days))
    registration_dates.append(random_date.date())

# Создаем датафрейм пользователей
users = pd.DataFrame({
    'user_id': user_ids,
    'registration_date': registration_dates,
    'country': np.random.choice(['USA', 'UK', 'Germany', 'France', 'India'], size=1000, p=[0.3, 0.2, 0.15, 0.15, 0.2]),
    'device': np.random.choice(['iPhone', 'Android'], size=1000, p=[0.6, 0.4])
})

users.to_csv('C:/Users/majkl/WorkSpace/Analysis_user_in_app/users.csv', index=False, encoding='utf-8')

print(f"Создан файл users.csv с {len(users)} записями")
```

```python
# Теперь создаем данные о событиях
event_names = ['app_launch', 'search', 'view_product', 'add_to_cart', 'purchase', 'logout']
events_list = []

# Для каждого пользователя генерируем поток событий
for user_id in user_ids:
    # Определяем дату регистрации этого пользователя
    reg_date = users[users['user_id'] == user_id]['registration_date'].iloc[0]
    reg_date = datetime.strptime(reg_date, '%Y-%m-%d') if isinstance(reg_date, str) else datetime.combine(reg_date, datetime.min.time())

    # Количество сессий у пользователя
    num_sessions = np.random.poisson(5) # В среднем 5 сессий

    for session in range(max(1, num_sessions)):
        session_time = reg_date + timedelta(days=np.random.randint(0, 90), hours=np.random.randint(0, 24), minutes=np.random.randint(0, 60))
        # В каждой сессии несколько событий
        session_events = np.random.choice(event_names, size=np.random.randint(2, 10), p=[0.3, 0.2, 0.2, 0.15, 0.1, 0.05])
        
        for event in session_events:
            session_time += timedelta(seconds=np.random.randint(10, 300)) # События идут с интервалом
            events_list.append({
                'user_id': user_id,
                'event_timestamp': session_time,
                'event_name': event
            })

events = pd.DataFrame(events_list)           
events.to_csv('C:/Users/majkl/WorkSpace/Analysis_user_in_app/events.csv', index=False, encoding='utf-8')

print(f"Создан файл users.csv с {len(users)} записями")
```
