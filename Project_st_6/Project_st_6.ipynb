## 2. Создайте новый признак – марку автомобиля (company). Машины каких производителей встречаются в датасете? Далее исправьте названия и проверьте изменения.

```python
cars['CarName'] = cars.CarName.apply(lambda x: x.split(' ')[0])
cars.head()
```

```python
cars.CarName.nunique()
```

```python
cars.CarName.unique()
```

```python
def clean_brand_names(line):
    brand = line["CarName"]
    if brand == "maxda":
        return "mazda"
    elif brand == "vokswagen" or brand == "vw":
        return "volkswagen"
    elif brand == "Nissan":
        return "nissan"
    elif brand == "porcshce":
        return "porsche"
    elif brand == "toyouta":
        return "toyota"
    else:
        return brand
```

```python
cars["CarName"] = cars.apply(lambda line: clean_brand_names(line), axis=1)
cars.CarName.nunique()
```

```python
cars.CarName.unique()
```

## 3. Посчитайте корреляцию между price и другими переменными.

```python
cars_2 = cars[['CarName', 'fueltype', 'aspiration', 'carbody', 'drivewheel', 'wheelbase', 'carlength', 'carwidth',
            'curbweight', 'enginetype', 'cylindernumber', 'enginesize', 'boreratio', 'horsepower', 'price']]
cars_2.head()
```

```python
cars_2.corr(numeric_only=True)
```

```python
cars_2.corr(numeric_only=True)['price'].sort_values(ascending=False).round(2)
```

## 4. Преобразуйте категориальные переменные с помощью pd.get_dummies(). 

```python
cars_2.dtypes
```

```python
df_dummy = pd.get_dummies(data=cars_2[['CarName', 'fueltype', 'aspiration', 'carbody', 'drivewheel', 'enginetype', \
                                       'cylindernumber']], drop_first = True)
df_dummy.head()
```

```python
cars_new = pd.concat([cars_2, df_dummy], axis=1)
cars_new = cars_new.drop(columns=['CarName', 'fueltype', 'aspiration', 'carbody', 'drivewheel', 'enginetype', 'cylindernumber'])
cars_new.head()
```

```python
cars_new.shape
```

## 5. Постройте модель с одним предиктором цены – horsepower. Какой процент изменчивости объясняет полученная модель? (\(R^2\))

```python
results = smf.ols('price ~ horsepower', cars_new).fit()
print(results.summary())
```

## 6. Далее – две модели (со всеми предикторами и со всеми, кроме марок машин). Обратите внимание на изменения в \(R^2\), коэффициентах и их значимости. Какую модель лучше оставить? 

Лучше оставить модель, где  меньше предикторов, ведь R2 изменился не очень сильно, а часть марок вообще не значима

```python
cars_new.info()
```

```python
x = cars_new.drop(['price'], axis='columns')
x = x[['wheelbase', 'carlength', 'carwidth', 'curbweight', 'enginesize', 'boreratio', 'horsepower']]
x = sm.add_constant(x)
y = cars_new['price']
```

```python
model_1 = sm.OLS(y, x).fit()
print(model_1.summary())
```

```python
cols_to_keep = cars_new.columns[~cars_new.columns.str.startswith('company_')]
cols_to_keep
```

```python
x = cars_new[cols_to_keep].drop('price', axis='columns')
x = x[['wheelbase', 'carlength', 'carwidth', 'curbweight', 'enginesize', 'boreratio', 'horsepower']]
x = sm.add_constant(x)
y = cars_new['price']
```

```python
model_2 = sm.OLS(y, x).fit()
print(model_2.summary())
```

Выбранная модель объясняет примерно 90% дисперсии (окр. до целого). Среди предикторов 10 из 27 оказались не значимыми (p > 0.05). Пример интерпретации: при единичном изменении показателя horsepower, цена ВОЗРАСТАЕТ на 86.8164 (без округления).
